```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
df.columns = df.columns.str.replace(
    r"\s+", " ", regex=True
).str.strip()

# Independent variables (X)
feature_names = [
    "Total Food Expenditure",
    "Imputed House Rental Value",
    "Education Expenditure",
    "Total Number of Family members",
    "Electricity"
]

# Dependent variable (y)
target_name = "Total Household Income"

# Select and clean our dataset
data = df[feature_names + [target_name]].copy()
data = data.apply(pd.to_numeric, errors="coerce")
data = data.dropna()

# Convert to NumPy arrays
X_raw = data[feature_names].to_numpy(dtype=float)
y = data[target_name].to_numpy(dtype=float)

# Display dimensions
n_samples, n_features = X_raw.shape

print("Number of samples:", n_samples)
print("Number of features:", n_features)
print("Feature names:", feature_names)

# Show descriptive statistics
display(pd.DataFrame(X_raw, columns=feature_names).describe())
```
